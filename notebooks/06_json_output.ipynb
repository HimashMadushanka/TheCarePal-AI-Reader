```python
import json
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
if "medicines" not in globals():
    medicines = json.loads((PROJECT_ROOT / "data" / "ground_truth" / "medicines.json").read_text(encoding="utf-8"))[0]["medicines"]
if "lab_values" not in globals():
    lab_values = json.loads((PROJECT_ROOT / "data" / "ground_truth" / "lab_values.json").read_text(encoding="utf-8"))[0]["lab_results"]
result = {"medicines": medicines, "lab_results": lab_values}
output_dir = PROJECT_ROOT / "outputs" / "json"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "extracted_result.json"
output_path.write_text(json.dumps(result, indent=2, ensure_ascii=False) + "\n", encoding="utf-8")
print(json.dumps(result, indent=2, ensure_ascii=False))
print(f"Saved to {output_path}")
```
