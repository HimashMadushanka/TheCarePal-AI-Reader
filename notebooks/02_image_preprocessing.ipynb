```python
from pathlib import Path
import cv2
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
image_files = sorted((PROJECT_ROOT / "data").rglob("*.png")) + sorted((PROJECT_ROOT / "data").rglob("*.jpg"))

def preprocess_image(image_path):
    image = cv2.imread(str(image_path))
    if image is None:
        raise ValueError(f"Cannot read image: {image_path}")
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    denoised = cv2.medianBlur(gray, 3)
    return cv2.adaptiveThreshold(denoised, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 31, 11)

if not image_files:
    print("No image samples yet. Add a PNG or JPG under data/ to try preprocessing.")
else:
    processed_image = preprocess_image(image_files[0])
    plt.figure(figsize=(12, 6))
    plt.imshow(processed_image, cmap="gray")
    plt.axis("off")
    plt.show()
```
