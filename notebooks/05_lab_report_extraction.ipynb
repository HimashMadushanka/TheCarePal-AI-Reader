```python
import re
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

def extract_lab_values(text):
    results = []
    pending_test = None
    for line in text.splitlines():
        line = line.strip()
        value_match = re.match(r"^(.+?):\s*([+-]?\d+(?:\.\d+)?)\s*([^\s]+)?$", line)
        range_match = re.match(r"^Reference Range:\s*(.+)$", line, re.I)
        if value_match:
            pending_test = {"test_name": value_match.group(1).strip(), "value": value_match.group(2), "unit": (value_match.group(3) or "").strip(), "reference_range": ""}
            results.append(pending_test)
        elif range_match and pending_test:
            pending_test["reference_range"] = range_match.group(1).strip()
            pending_test = None
    return results

text = (PROJECT_ROOT / "data" / "lab_reports" / "lab_01.txt").read_text(encoding="utf-8")
lab_values = extract_lab_values(text)
for result in lab_values:
    print(result)
```
