```python

import re

def extract_lab_values(text):

    lab_results = []

    for line in text.splitlines():

        line = line.strip()

        # Example: Hemoglobin 13.5 g/dL
        match = re.search(
            r"^(.+?)\s+(\d+(?:\.\d+)?)\s*([a-zA-Z/%]+)(?:\s+(\d+(?:\.\d+)?\s*[-–]\s*\d+(?:\.\d+)?))?$",
            line
        )

        if match:

            lab_results.append({
                "test_name": match.group(1).strip(),
                "value": match.group(2),
                "unit": match.group(3),
                "reference_range": match.group(4) or ""
            })

    return lab_results


lab_values = extract_lab_values(text)

for result in lab_values:
    print(result)
```
