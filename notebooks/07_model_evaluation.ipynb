```python

import pandas as pd
from sklearn.metrics import accuracy_score

df = pd.read_csv(
    "../data/ground_truth/evaluation.csv"
)

results = []

for field, group in df.groupby("field"):

    accuracy = accuracy_score(
        group["expected"].astype(str),
        group["predicted"].astype(str)
    )

    results.append({
        "Field": field,
        "Accuracy": round(accuracy * 100, 2)
    })

evaluation = pd.DataFrame(results)

print(evaluation)
```
