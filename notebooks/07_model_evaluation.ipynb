```python
import json
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
truth_path = PROJECT_ROOT / "data" / "ground_truth" / "medicines.json"
truth = json.loads(truth_path.read_text(encoding="utf-8"))
prediction_path = PROJECT_ROOT / "outputs" / "predictions.json"
if not prediction_path.exists():
    print("No predictions.json found yet. Save model predictions there to calculate accuracy against the medicine ground truth.")
else:
    predictions = json.loads(prediction_path.read_text(encoding="utf-8"))
    print("Ground-truth samples:", len(truth))
    print("Prediction samples:", len(predictions))
    print("Add field-level expected/predicted rows to compare detailed accuracy.")
```
