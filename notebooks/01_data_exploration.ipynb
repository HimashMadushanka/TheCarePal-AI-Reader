```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
PRESCRIPTION_DIR = PROJECT_ROOT / "data" / "prescriptions"
LAB_DIR = PROJECT_ROOT / "data" / "lab_reports"
prescriptions = sorted(p.name for p in PRESCRIPTION_DIR.glob("*.txt"))
lab_reports = sorted(p.name for p in LAB_DIR.glob("*.txt"))
print("Prescription documents:", len(prescriptions))
print("Lab reports:", len(lab_reports))
print("Prescription samples:", prescriptions)
print("Lab report samples:", lab_reports)
```
